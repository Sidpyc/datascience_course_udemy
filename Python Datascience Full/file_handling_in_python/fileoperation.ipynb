```python
### Read a file

with open("example.txt",'r') as file:
    content = file.read()
    print(content)
```

```python
with open('example.txt', 'r') as file: 
    for line in file:
        print(line.strip()) #.strip() removes the new line / spaces
```

```python
## Overwriting a file

with open('example.txt', 'w') as file:
    file.write("Hello World\n")
    file.write("This is a new line")
```

```python
## appending to a file

with open ('example.txt', 'a') as file:
    file.write("\nThis is another line")
```

```python
#Writing a list of lines

lines=['\nFirst Line\n', 'Second Line\n', 'Third Line\n']
with open('example.txt', 'a') as file:
    file.writelines(lines) 
```

```python
## Read the content from a source file to destination file
with open ('example.txt', 'r') as file:
    content = file.read()

with open ('destination.txt','w') as dest_file:
    dest_file.write(content)
```

```python
# Read a file and count number of lines, words, and characters.

with open('example.txt', 'r') as file:
    lines=file.readlines()
    line_count=len(lines)
    word_count = sum(len(line.split()) for line in lines)
    char_count = sum(len(line) for line in lines)

print(f"lines:{line_count}, words:{word_count}, characters:{char_count}")
```

```python
# Writing and then reading a file

with open('example.txt', 'w+') as file:
    file.write("Hello world\n")
    file.write('This is a new line \n')

    file.seek(0)
    content = file.read()

    print(content)
```
