```python
class Stack:
    def __init__(self):
        self.stack = []

    def push(self,element):
        self.stack.append(element)
           
    def pop(self):
        if not self.isEmpty():
            ele = self.stack[-1]
            self.stack.pop(-1)
            print(f"{ele} has been removed")
        
        else:
            print('Stack Empty')
    
    def isEmpty(self):
        return len(self.stack)==0

    def peek(self):
        if not self.isEmpty():
            print( f"Latest Element: {self.stack[-1]}")
        else:
            print("Stack Empty")

    def size(self):
        return f'Size :{len(self.stack)}'

    def display(self):
        for i in self.stack:
            print(f"|{i}|")



new_stack = Stack()
new_stack.push(2)
new_stack.push(3)
new_stack.push(4)
new_stack.display()
print(new_stack.size())
print(new_stack.isEmpty())
new_stack.peek()
new_stack.pop()
new_stack.peek()
new_stack.display()


```
